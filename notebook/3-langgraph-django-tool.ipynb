```python
from ai.tools import get_documents_tool, get_document_tool
from django.contrib.auth.models import User

user_id = User.objects.filter(is_superuser=True).first().id

config = {
    "configurable": {
        "user_id": user_id,
    }
}

# 1st arg = tool input, 2nd arg = config
get_documents_tool.invoke({}, config)
get_document_tool.invoke({"document_id": 2}, config)
```
