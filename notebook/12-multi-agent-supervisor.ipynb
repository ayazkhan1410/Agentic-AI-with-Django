```python
from ai.supervisor import get_supervisor

supervisor = get_supervisor()
```

```python
import uuid
from django.contrib.auth.models import User

config = {
    "configurable": {
        "user_id": User.objects.filter(is_superuser=True).first().id,
        "thread_id": uuid.uuid4(),
    }
}

# 1st arg = agent input 
# 2nd arg = config (user_id goes here)
response = supervisor.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Please Delete any duplicate documents"
            }
        ],
    },
    config,
)

# Print only the final reply (avoids huge collapsed notebook output)
print(response["messages"][-1].content)
```
